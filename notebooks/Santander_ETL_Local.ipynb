## Extract

Leitura dos clientes fictícios a partir de um arquivo CSV.

```python
import pandas as pd

clientes = pd.read_csv('../data/clientes.csv')
clientes.head()
```

## Transform

Criação de mensagens personalizadas para cada cliente.

```python
def gerar_mensagem(row):
    nome = row['nome']
    perfil = str(row['perfil']).lower()
    objetivo = str(row['objetivo']).lower()

    if 'conservador' in perfil:
        return f'{nome}, comece com segurança: sua reserva fortalece seus próximos investimentos.'
    if 'arrojado' in perfil:
        return f'{nome}, diversifique seus investimentos e transforme estratégia em crescimento.'
    if 'iniciante' in perfil:
        return f'{nome}, organize hoje seu dinheiro e dê o primeiro passo para investir melhor.'
    if 'viagem' in objetivo:
        return f'{nome}, invista aos poucos e aproxime sua próxima viagem da realidade.'

    return f'{nome}, investir todo mês pode aproximar você dos seus objetivos financeiros.'

clientes['news_description'] = clientes.apply(gerar_mensagem, axis=1)
clientes['news_icon'] = 'https://digitalinnovationone.github.io/santander-dev-week-2023-api/icons/credit.svg'
clientes.head()
```

## Load

Salvando o resultado em CSV e JSON.

```python
clientes.to_csv('../output/clientes_com_mensagens.csv', index=False, encoding='utf-8-sig')
clientes.to_json('../output/clientes_com_mensagens.json', orient='records', force_ascii=False, indent=2)
print('Arquivos gerados com sucesso!')
```
